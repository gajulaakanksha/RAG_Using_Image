```python
%%writefile app_1.py
import streamlit as st
import google.generativeai as genai
from PIL import Image
import io
import base64

# Configure Google Gemini API
genai.configure(api_key="Gemini_API_Key")

# Function to convert image to Base64 format
def encode_image(image):
    img_byte_arr = io.BytesIO()
    image.save(img_byte_arr, format=image.format)
    return base64.b64encode(img_byte_arr.getvalue()).decode('utf-8')

# Function to process image with Gemini 2.0 Flash
def analyze_image_gemini_2(image):
    model = genai.GenerativeModel('gemini-2.0-flash')
    image_data = {
        "mime_type": "image/png",  # Adjust MIME type as per your image format
        "data": encode_image(image)
    }
    response = model.generate_content([{"inline_data": image_data}])
    return response.text if response else "No response from model."

# Function to process image with Gemini 1.5 Flash
def analyze_image_gemini_1_5(image):
    model = genai.GenerativeModel('gemini-1.5-flash')
    image_data = {
        "mime_type": "image/png",  # Adjust MIME type as per your image format
        "data": encode_image(image)
    }
    response = model.generate_content([{"inline_data": image_data}])
    return response.text if response else "No response from model."

# Streamlit UI
st.title("Upload an Image for Analysis with Gemini 2.0 & 1.5")

# File uploader
uploaded_file = st.file_uploader("Choose an image...", type=["jpg", "png", "jpeg"])

if uploaded_file:
    # Show uploaded image
    image = Image.open(uploaded_file)
    st.image(image, caption="Uploaded Image", use_column_width=True)

    # Process image with Gemini 2.0 Flash
    st.subheader("Analysis using Gemini 2.0 Flash:")
    response_2 = analyze_image_gemini_2(image)
    st.write(response_2)

    # Process image with Gemini 1.5 Flash
    st.subheader("Analysis using Gemini 1.5 Flash:")
    response_1_5 = analyze_image_gemini_1_5(image)
    st.write(response_1_5)



```
